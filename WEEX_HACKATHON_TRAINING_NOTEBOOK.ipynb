```python
# 2. Configure Python Path
import sys
import os

# Assuming you uploaded the repo as a dataset or files
# If uploaded plainly, current dir is fine.
# We print cwd to check.
print(f"Current Dir: {os.getcwd()}")
sys.path.append(os.getcwd()) # Add root to path
```

```python
# 3. Get Data (Direct Download)
# Since WEEX API has limits, we use Yahoo Finance for Deep History
import yfinance as yf

SYMBOL = "cmt_btcusdt"
print(f"📥 Downloading Deep History for {SYMBOL}...")

# Map to YF Symbol (BTC-USD)
data = yf.download("BTC-USD", period="6mo", interval="15m")
data.reset_index(inplace=True)

# Standardize Columns for Bot (lowercase)
if 'Datetime' in data.columns: data.rename(columns={"Datetime": "timestamp"}, inplace=True)
if 'Date' in data.columns: data.rename(columns={"Date": "timestamp"}, inplace=True)

# Map other columns if needed (YF sometimes returns MultiIndex)
# Simple fix for flattening YF columns:
data.columns = [c[0] if isinstance(c, tuple) else c for c in data.columns]
data.columns = [c.lower() for c in data.columns]

# Save to match expected path
os.makedirs("data", exist_ok=True)
CSV_PATH = "data/BTC_USDT_15m.csv"
data.to_csv(CSV_PATH, index=False)
print(f"✅ Saved 6 months of data ({len(data)} rows) to {CSV_PATH}")
```

```python
# 4. Run Training (BTC)
# Run the script
!python scripts/train_dqn.py --symbol {SYMBOL} --data {CSV_PATH} --episodes 50 --tune 10
```
